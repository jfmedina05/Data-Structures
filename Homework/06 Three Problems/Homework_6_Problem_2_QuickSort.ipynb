# Homework 6 - Problem 2: QuickSort

## Objective

The goal of this problem is to sort a list of numbers using the **QuickSort algorithm**.

QuickSort is a recursive divide-and-conquer sorting algorithm.

## How QuickSort Works

QuickSort follows these steps:

1. Select a number from the list to use as the **pivot**.
2. Place numbers less than or equal to the pivot into a left list.
3. Place numbers greater than the pivot into a right list.
4. Recursively apply QuickSort to the left and right lists.
5. Combine the sorted left list, the pivot, and the sorted right list.

The recursion stops when a list has zero or one element because that list is already sorted.

```python
def quicksort(numbers):
    # Base case: an empty list or one-item list is already sorted
    if len(numbers) <= 1:
        return numbers

    # Choose the first element as the pivot
    pivot = numbers[0]

    # Numbers less than or equal to the pivot
    left = [x for x in numbers[1:] if x <= pivot]

    # Numbers greater than the pivot
    right = [x for x in numbers[1:] if x > pivot]

    # Recursively sort both sides and combine the result
    return quicksort(left) + [pivot] + quicksort(right)
```

## Test the QuickSort Program

The following unsorted list will be passed into the QuickSort function.

```python
numbers = [33, 2, 3, 45, 6, 54, 33]

print("Original list:")
print(numbers)

sorted_numbers = quicksort(numbers)

print("\nSorted list:")
print(sorted_numbers)
```

## Example of the Recursive Process

For the list:

`[33, 2, 3, 45, 6, 54, 33]`

the first pivot is `33`.

The list is divided into:

- Left side: `[2, 3, 6, 33]`
- Pivot: `[33]`
- Right side: `[45, 54]`

QuickSort then sorts the left and right sides recursively before combining them.

## Conclusion

QuickSort sorts the list by repeatedly dividing it into smaller lists.

Recursion continues until the smaller lists contain zero or one element. The sorted pieces are then combined to create the final sorted list.
