# Homework 6 - Problem 3: Pig Latin Using Regular Expressions

## Objective

The goal of this problem is to translate English text into **Pig Latin** using Python regular expressions.

Python's `re` module will be used to recognize parts of each word and apply the appropriate Pig Latin rule.

## Pig Latin Rules

This solution follows the rules demonstrated by the provided Perl solution.

### Rule 1: Word begins with a vowel

If a word begins with a vowel, keep the word in the same order and add `way`.

Example:

`is` → `isway`

### Rule 2: Word begins with a consonant or consonant cluster

Move the consonant or consonant cluster from the beginning of the word to the end and add `ay`.

Examples:

`Hi` → `iHay`

`What` → `atWhay`

`good` → `oodgay`

### Rule 3: Punctuation

Punctuation at the beginning or end of a word should remain in its original position.

Example:

`now.` → `ownay.`

### Rule 4: Non-word values

Values that do not contain a regular English word, such as `4,000`, are left unchanged.

## Import the Regular Expression Module

Python's `re` module provides the regular expression functions used to match and modify each word.

```python
import re
```

## Translate One Word

The function below uses regular expressions to divide each token into:

- punctuation before the word,
- the letters in the word,
- and punctuation after the word.

It then checks whether the word begins with a vowel or a consonant cluster.

```python
def translate_word(token):
    # Separate beginning punctuation, letters, and ending punctuation
    match = re.fullmatch(
        r'([^A-Za-z]*)([A-Za-z]+)([^A-Za-z]*)',
        token
    )

    # If the token is not a normal word, leave it unchanged
    if not match:
        return token

    prefix, word, suffix = match.groups()

    # If the word begins with a vowel, add "way"
    if re.match(r'^[AEIOUaeiou]', word):
        translated = word + "way"

    # Otherwise, find the consonant cluster at the beginning
    else:
        consonant_match = re.match(
            r'^([^AEIOUaeiou]+)(.*)$',
            word
        )

        consonants, rest = consonant_match.groups()

        translated = rest + consonants + "ay"

    # Put punctuation back in its original location
    return prefix + translated + suffix
```

## Translate a Complete Message

The next function finds every non-space token in the message and passes it to the Pig Latin translation function.

Using `re.sub()` allows the regular expression to process the entire sentence while keeping the original spaces and line breaks.

```python
def pig_latin(text):
    return re.sub(
        r'\S+',
        lambda match: translate_word(match.group()),
        text
    )
```

## Test Using the Example From the Assignment

```python
text = """Hi I am here now.
What is going on?
Are you having a reasonable good Friday?"""

print("Original text:")
print(text)

print("\nPig Latin:")
print(pig_latin(text))
```

## Explanation of the Regular Expressions

The program uses several regular expressions.

`([^A-Za-z]*)([A-Za-z]+)([^A-Za-z]*)`

separates a token into beginning punctuation, the actual word, and ending punctuation.

`^[AEIOUaeiou]`

checks whether the word begins with a vowel.

`^([^AEIOUaeiou]+)(.*)$`

separates the consonant or consonant cluster at the beginning of a word from the rest of the word.

Finally, `re.sub()` applies the translation function to each token in the complete message.
