## **Problem 1 - Complex Number Class**

This section implements a custom `Complex` class for working with complex numbers in the form:

\[
a + bi
\]

where `a` is the real part and `b` is the imaginary part.

The class uses the `__init__()` method to store the real and imaginary values for each object. It also uses Python operator overloading so that `Complex` objects can be added and multiplied using the normal `+` and `*` operators.

- `__add__()` adds the real parts together and the imaginary parts together.
- `__mul__()` multiplies two complex numbers using the standard complex-number multiplication formula.
- `__str__()` controls how the complex number is displayed when printed.

For example, if:

\[
a = 2 + 3i
\]

and

\[
b = 4 + 5i
\]

then the program can calculate and display both `a + b` and `a * b`.

This exercise demonstrates classes, objects, instance attributes, special methods, and operator overloading in Python.

```python
class Complex(object):

    def __init__(self, real, imaginary):
        self.real = real
        self.imaginary = imaginary

    def __add__(self, other):
        return Complex(
            self.real + other.real,
            self.imaginary + other.imaginary
        )

    def __mul__(self, other):
        return Complex(
            self.real * other.real - self.imaginary * other.imaginary,
            self.real * other.imaginary + self.imaginary * other.real
        )

    def __str__(self):
        return str(self.real) + " + " + str(self.imaginary) + "i"


a = Complex(2, 3)
b = Complex(4, 5)

print("a =", a)
print("b =", b)
print("a + b =", a + b)
print("a * b =", a * b)
```

## **Problem 2 - Creature Inheritance**

This section implements a base `Creature` class along with `Dog` and `Cow` subclasses.

The `Creature` class stores a creature's name using the `__init__()` method and includes a `speak()` method that describes the creature.

`Dog` and `Cow` inherit from `Creature`, which means they automatically receive the name attribute from the parent class. Each subclass then overrides the `speak()` method to provide its own behavior.

- `Creature` prints a general creature message.
- `Dog` prints the creature's name and adds `woof`.
- `Cow` prints the creature's name and adds `moo`.

For example, creating a `Dog` object named `Buddy` allows the program to call:

`Buddy.speak()`

and produce a dog-specific message.

This exercise demonstrates classes, inheritance, parent and child classes, instance attributes, and method overriding in Python.

```python
class Creature(object):

    def __init__(self, name):
        self.name = name

    def speak(self):
        print("My name is " + self.name + " and I am Creature.")


class Dog(Creature):

    def speak(self):
        print("My name is " + self.name + " and I am a Dog: woof.")


class Cow(Creature):

    def speak(self):
        print("My name is " + self.name + " and I am a Cow: moo.")


creature = Creature("Alex")
dog = Dog("Buddy")
cow = Cow("Bessie")

creature.speak()
dog.speak()
cow.speak()
```
