## **Problem 4**

This problem uses a weighted sample space where each possible sum of eight die rolls is mapped to the number of ways that sum can occur.

The `is_in_interval` function checks whether a sum is between `14` and `24` inclusive. The probability that eight rolls produce a sum in this interval is:

**0.23694701646090535, or about 23.69%.**

This uses interval analysis to calculate the probability of a range of values instead of one exact value.


```python
def get_matching_event(event_condition, sample_space):
    return set([outcome for outcome in sample_space if event_condition(outcome)])

def compute_event_probability(event_condition, generic_sample_space):
    event = get_matching_event(event_condition, generic_sample_space)

    if type(generic_sample_space) == type(set()):
        return len(event) / len(generic_sample_space)

    event_size = sum(generic_sample_space[outcome] for outcome in event)
    return event_size / sum(generic_sample_space.values())
```

```python
possible_rolls = list(range(1, 7))
print(possible_rolls)
```

```python
from itertools import product
from collections import defaultdict

sample_space = set(product(possible_rolls, repeat=8))

weighted_sample_space = defaultdict(int)

for outcome in sample_space:
    total = sum(outcome)
    weighted_sample_space[total] += 1
```

```python
def is_in_interval(number, minimum, maximum):
    return minimum <= number <= maximum

def is_in_interval(number, minimum, maximum):
    return minimum <= number <= maximum

prob = compute_event_probability(
    lambda x: is_in_interval(x, 14, 24),
    weighted_sample_space
)

print(f"Probability of interval is {prob}")
print(f"Percentage: {prob * 100:.2f}%")
```
