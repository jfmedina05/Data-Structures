## What basic Matplotlib plots will we be using?

In this notebook, we will look at the basic Matplotlib plotting tools introduced in Chapter 2 of *Data Science Bookcamp*.

The main tools we will use are:

- `plt.plot()` for line plots
- `plt.scatter()` for scatter plots
- `plt.fill_between()` for shading an area underneath a plot
- `plt.xlabel()` and `plt.ylabel()` for labeling the axes

## Importing Matplotlib

Matplotlib is a Python library used for creating graphs and visualizations.

We import the `pyplot` module using the common abbreviation `plt`.

```python
import matplotlib.pyplot as plt
```

## Creating Example Data

To practice using Matplotlib, we will create two variables.

`x` contains the numbers from 0 through 9.

`y` contains twice the value of each number in `x`.

```python
x = range(0, 10)
y = [2 * value for value in x]

print(list(x))
print(y)
```

## Line Plot

The `plt.plot()` function plots the x-values against the y-values and connects the points using lines.

```python
plt.plot(x, y)
plt.show()
```

## Scatter Plot

The `plt.scatter()` function displays each coordinate as an individual point instead of connecting the points with lines.

```python
plt.scatter(x, y)
plt.show()
```

## Shading an Interval

The `plt.fill_between()` function can highlight an area underneath a curve.

First, we create a function that checks whether a value falls inside an interval. We will use it to highlight the region where x is between 2 and 6.

```python
def is_in_interval(val, start, end):
    return start <= val <= end
```

```python
plt.plot(x, y)

where = [is_in_interval(value, 2, 6) for value in x]

plt.fill_between(x, y, where=where)

plt.show()
```

The shaded region shows the portion of the graph where the x-values fall between 2 and 6.

## Combining the Plotting Tools

We can combine a line plot, scatter plot, and shaded region in the same graph.

This lets us see the individual data points, the line connecting them, and a highlighted interval at the same time.

```python
plt.scatter(x, y)
plt.plot(x, y)
plt.fill_between(x, y, where=where)

plt.show()
```

## Labeling the Axes

Graphs are easier to understand when the x-axis and y-axis have labels.

Matplotlib provides:

- `plt.xlabel()` for the x-axis
- `plt.ylabel()` for the y-axis

```python
plt.plot(x, y)

plt.xlabel('Values between zero and ten')
plt.ylabel('Twice the values of x')

plt.show()
```
