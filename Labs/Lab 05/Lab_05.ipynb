### Question 2 — Minimum Number of Dimensions

The `ndmin` argument tells NumPy the minimum number of dimensions the new array must have.

- `what(1)` creates a **1-D** array: `[1 2 3 4]`
- `what(2)` creates a **2-D** array: `[[1 2 3 4]]`
- `what(3)` creates a **3-D** array: `[[[1 2 3 4]]]`

The data stays the same, but NumPy adds additional axes around it. The `a.ndim` value therefore prints `1`, `2`, and `3`, respectively.

```python
#  2. consider this function:

def what(n):
  a = np.array([1, 2, 3, 4], ndmin=n)
  print(a.ndim)
  return a

#     what is the difference between what(1), what(2) and what(3)

print(what(1)) # what do these print and why?
print(what(2))
print(what(3))
```

### Question 14 — String Arrays and Type Conversion

Using `dtype='S'` stores the values as fixed-width byte strings instead of numbers.

- `a.dtype` prints a string data type such as `|S1`.
- `a[0] + a[1]` produces `b'12'` because the byte strings `b'1'` and `b'2'` are concatenated.
- `a.astype('i')` converts the string representations of the numbers into integers.
- `b[0] + b[1] + b[3]` performs numerical addition: `5 + 1 + 3 = 9`.
- `a[0] + a[1] + a[3]` performs byte-string concatenation and produces `b'513'`.

This demonstrates that the data type determines how operations such as `+` behave.

```python
# 14. insert the correct argument to specify that array should be of type STRING

a = np.array([1, 2, 3, 4], dtype="S")

#     then predict and explain the output for the following expressions

print(a.dtype)
print(a[0] + a[1])
a = np.array([5, 1, -1, 3], dtype='S')
b = a.astype('i') # what happens here?
print(b[0] + b[1] + b[3])
print(a[0] + a[1] + a[3])
```

### Question 16 — Copy vs. Assignment

`a.copy()` creates an independent copy of the original array.

Therefore, changing `x[0]` to `-5` does **not** change `a`.

However, `b = a` does not make a new array. Both variables refer to the same array object, so changing `b[0]` to `-25` also changes `a`.

Final output:

`[1 2 3 4 5]`

followed by:

`[-25 2 3 4 5]`

```python
# 16. use the correct method to make a copy of the array

a = np.array([1, 2, 3, 4, 5])
x = a.copy() # must use a to make a copy of it

#     then predict the effect of the following expressions

x[0] = -5
print(a) # does a see the change?
b = a
b[0] = -25
print(a) # does a see the change?
```

### Question 17 — NumPy Views and the `base` Attribute

A view shares its underlying data with another NumPy array.

- `a.base` is `None` because `a` owns its data.
- `d.base` refers to `a` because `d = a.view()` shares `a`'s data.
- `b = a` is simply another reference to the same array, so `b.base` is also `None`.
- `c = a.copy()` creates independent data, so `c.base` is `None`.

The view is different because it is a separate NumPy array object that still uses the original array's data.

```python
# 17. use the correct method to make a view of this array:

a = np.array([1, 2, 3, 4, 5])
d = a.view()

#     then predict the effect of these expressions:

print(a.base)
print(d.base) # why is this different from all the others?
b = a
print(b.base)
c = a.copy()
print(c.base)
```

## Part 4 — Array Shapes, Reshaping, and Joining

The shape of an array describes how its elements are organized. NumPy can reorganize the same elements into different dimensions as long as the total number of elements stays the same.

### Question 18 — Shape and `reshape()`

The original array has four elements in one dimension, so:

`a.shape` → `(4,)`

An important detail is that `reshape()` **returns a reshaped array but does not permanently change `a` unless the result is assigned to a variable**.

Because none of the reshape results are saved, every `print(a)` still prints:

`[1 2 3 4]`

For example, `a = a.reshape(2, 2)` would be required to permanently store the new 2-D shape.

```python
# 18. predict (and explain) the outcome of these expressions

a = np.array([1, 2, 3, 4])

print(a.shape)

a.reshape(1, 4)
print(a)

a.reshape(2, 2)
print(a)

a.reshape(4, 1)
print(a)
```

### Question 24 — Exercise 3-1: Maximum After-Tax Income

First, the after-tax income is calculated element-by-element using:

`salaries - salaries * taxation`

The maximum value in the resulting matrix is `81.0`.

The expression `after_tax_income == max_income` creates a Boolean matrix that is `True` only where the maximum occurs. `np.nonzero()` finds the row containing that value, and that row index is then used to select the corresponding name.

The person with the highest after-tax income is **Tim**, with an after-tax income of **$81,000**.

```python
# 24. solve exercise 3-1 on page 57

import numpy as np

# Data: yearly salary in ($1000) [2017, 2018, 2019]
alice = [99, 101, 103]
bob = [110, 108, 105]
tim = [90, 88, 85]

names = np.array(['Alice', 'Bob', 'Tim'])

salaries = np.array([alice, bob, tim])

taxation = np.array([
    [0.2, 0.25, 0.22],
    [0.4, 0.5, 0.5],
    [0.1, 0.2, 0.1]
])

# Calculate each person's after-tax income
after_tax_income = salaries - salaries * taxation

# Find the maximum after-tax income
max_income = np.max(after_tax_income)

# Use selective indexing to find the person associated with that maximum
max_name = names[np.nonzero(after_tax_income == max_income)[0]]

print("Maximum after-tax income:", max_income)
print("Person with the maximum after-tax income:", max_name)
```

### Question 25 — Averaging Along Different Axes

The `axis` argument determines which dimension NumPy collapses while calculating the average.

**`axis=0`**

NumPy averages corresponding positions across the two outer arrays:

`[[3. 4.]`
` [5. 6.]]`

**`axis=1`**

NumPy averages across the rows inside each outer array:

`[[2. 3.]`
` [6. 7.]]`

**`axis=2`**

NumPy averages the values inside each innermost row:

`[[1.5 3.5]`
` [5.5 7.5]]`

Changing the axis changes the direction in which the multidimensional array is collapsed.

```python

# 25. predict and explain the effect of the following expressions

a = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])
print(np.average(a, axis=0)) # review the meaning of the axis parameter
print(np.average(a, axis=1))
print(np.average(a, axis=2))

print(a)
```
