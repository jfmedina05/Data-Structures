## **Problem 3**

This problem creates the sample space for eight rolls of a fair six-sided die. Each outcome contains eight numbers between `1` and `6`.

The event condition selects only the outcomes whose values add up to `28`. Using `compute_event_probability`, the probability is:

**0.08094350137174211, or about 8.09%.**

```python
def get_matching_event(event_condition, sample_space):
    return set([outcome for outcome in sample_space if event_condition(outcome)])

def compute_event_probability(event_condition, generic_sample_space):
    event = get_matching_event(event_condition, generic_sample_space)

    if type(generic_sample_space) == type(set()):
        return len(event) / len(generic_sample_space)

    event_size = sum(generic_sample_space[outcome] for outcome in event)
    return event_size / sum(generic_sample_space.values())
```

```python
# Define all of the possible rolls of the die
possible_rolls = list(range(1, 7))
print(possible_rolls)
```

```python
# Sample Space
from itertools import product

sample_space = set(product(possible_rolls, repeat=8))

len(sample_space)
```

```python
def has_sum_of_28(outcome):
    return sum(outcome) == 28

a = compute_event_probability(has_sum_of_28, sample_space)

print(f"8 rolls sum to 28 with a probability of {a}")
print(f"Percentage: {a * 100:.2f}%")

b = compute_event_probability(lambda x: sum(x) == 28, sample_space)

assert a == b
```
