## **Problem 1**

This problem defines sample spaces for both a fair and biased coin. For the fair coin, heads and tails are equally likely, so each has a probability of `0.5`.

For the biased coin, tails is weighted three times more than heads, giving:

- Heads: `0.25`
- Tails: `0.75`

The `compute_event_probability` function works with both regular and weighted sample spaces to calculate the probability of heads, tails, either outcome, or neither outcome.


```python
sample_space = {'Heads', 'Tails'}
```

```python
def is_heads_or_tails(outcome):
    return outcome in {'Heads', 'Tails'}

def is_neither(outcome):
    return not is_heads_or_tails(outcome)

def is_heads(outcome):
    return outcome == 'Heads'

def is_tails(outcome):
    return outcome == 'Tails'

def get_matching_event(event_condition, sample_space):
    return set([outcome for outcome in sample_space if event_condition(outcome)])
```

```python
event_conditions = [is_heads_or_tails, is_heads, is_tails, is_neither]
weighted_sample_space = {'Heads': 1, 'Tails': 3}
```

```python
def compute_event_probability(event_condition, generic_sample_space):
    event = get_matching_event(event_condition, generic_sample_space)

    if type(generic_sample_space) == type(set()):
        return len(event) / len(generic_sample_space)

    event_size = sum(generic_sample_space[outcome] for outcome in event)
    return event_size / sum(generic_sample_space.values())
```

```python
print("Biased coin results (tails is 3 times more likely than heads):")

for event_condition in event_conditions:
    prob = compute_event_probability(event_condition, weighted_sample_space)
    name = event_condition.__name__

    print(f"{name}: probability = {prob} ({prob * 100:.2f}%)")


print("\nFair coin results (heads and tails are equally likely):")

for event_condition in event_conditions:
    prob = compute_event_probability(event_condition, sample_space)
    name = event_condition.__name__

    print(f"{name}: probability = {prob} ({prob * 100:.2f}%)")
```
