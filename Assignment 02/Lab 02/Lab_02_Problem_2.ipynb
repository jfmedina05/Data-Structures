## **Problem 2**

This problem creates every possible combination of four children using `product`. Since each child is equally likely to be a boy or a girl, there are `16` equally likely outcomes.

The event condition selects the outcomes containing exactly two girls. There are `6` matching outcomes, so the probability is:

**0.375, or 37.5%.**

```python
def get_matching_event(event_condition, sample_space):
    return set([outcome for outcome in sample_space if event_condition(outcome)])

def compute_event_probability(event_condition, generic_sample_space):
    event = get_matching_event(event_condition, generic_sample_space)

    if type(generic_sample_space) == type(set()):
        return len(event) / len(generic_sample_space)

    event_size = sum(generic_sample_space[outcome] for outcome in event)
    return event_size / sum(generic_sample_space.values())

possible_children = ['Boy', 'Girl']
```

```python
from itertools import product

sample_space_efficient = set(product(possible_children, repeat=4))

sample_space_efficient
```

```python
def has_two_girls(outcome):
    return len([child for child in outcome if child == 'Girl']) == 2

prob = compute_event_probability(has_two_girls, sample_space_efficient)

print(f"Probability of 2 girls is {prob}")
print(f"Percentage: {prob * 100:.2f}%")
```
